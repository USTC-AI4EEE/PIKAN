```python
mit_data = torch.load('../Data/MIT_Data_Var2.pt')
hust_data = torch.load('../Data/HUST_Data_Var2.pt')
```

```python
mit_dataset = {
        'train_input': mit_data['train_input'],
        'train_label': mit_data['train_label'],
        'test_input': mit_data['test_input'],
        'test_label': mit_data['test_label']
    }
mit_dataset_cpu = {
        'train_input': mit_data['train_input'].to('cpu'),
        'train_label': mit_data['train_label'].to('cpu'),
        'test_input': mit_data['test_input'].to('cpu'),
        'test_label': mit_data['test_label'].to('cpu')
    }
hust_dataset = {
        'train_input': hust_data['train_input'],
        'train_label': hust_data['train_label'],
        'test_input': hust_data['test_input'],
        'test_label': hust_data['test_label']
    }
hust_dataset_cpu = {
        'train_input': hust_data['train_input'].to('cpu'),
        'train_label': hust_data['train_label'].to('cpu'),
        'test_input': hust_data['test_input'].to('cpu'),
        'test_label': hust_data['test_label'].to('cpu')
    }
```

```python
add_symbolic('sigmoid', lambda x: torch.sigmoid(x), c=4)
add_symbolic('relu', lambda x: torch.relu(x), c=3)
add_symbolic('cosh',torch.cosh)
add_symbolic('sinh',torch.sinh)
add_symbolic('arcsinh',torch.arcsinh)
add_symbolic('arccosh',torch.arccosh)

from sympy import symbols
from kan.utils import sigmoid,relu
from kan.compiler import expr2kan,ReLU,Sigmoid

sigmoid = sympy.Function('sigmoid')
relu = sympy.Function('relu')

x, t = symbols('x t')  # 合并定义变量，避免重复

expr = 173.545654296875 - 172.353 * cosh(
    0.288 * cosh(0.019 - 0.159 * relu(6.536*x - 0.992))  # 用符号表达式替换数值函数
    + 0.670 * cosh(-0.841 * sigmoid(5.688 - 6.370*x)  # 替换为符号表达式
                    - 0.044 * sinh(-1.171*t - 0.217) + 0.378) 
    - 1.075
)

# 步骤5：转换为模型（假设 expr2kan 是自定义函数，用于符号表达式转模型）
input_variables = (x, t)  # 明确输入变量
model = expr2kan(input_variables, expr,grid=3)

# 步骤6：使用模型（示例）
model.get_act(hust_dataset)  # 假设 hust_dataset 是输入数据
```
